**Question 1.1 (5 marks): Outlier Detection Using the IQR Method. Provide a brief description of the complete, step by step procedure in bullet point form.**

- We first need to remove any duplicate records, alongside remove any null records, these would heavily impact the IQR
- We then sort the dataset 
- To find IQR we need Q3 and Q1 
- Q3 is the 75th percentile at position (n * 0.75)
- Q1 is the 25th percentile at position (n * 0.25)
- IQR = Q3 - Q1
- Then we need to find the upper and lower limit using the IQR
- upper limit is Q3 + 1.5 * IQR
- lower limit is Q1 - 1.5 * IQR
- We then compare the upper and lower limit with the values below and above Q1 and Q3 respectivily to find outliers 



**Question 1.2 (5 marks): Python automation. Provide your Python code and include the required outputs.**

```python
### Write your answer here (code cell(s) to be used, as appropriate)
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt

dfHome = pd.read_csv("DataSet-HomeC.csv")
dfHome = dfHome[dfHome["use [kW]"].notna()]
dfHome = dfHome.sort_values(by="use [kW]", ascending=True)

Q3 = np.percentile(dfHome["use [kW]"], 75)
Q2 = np.percentile(dfHome["use [kW]"], 50)
Q1 = np.percentile(dfHome["use [kW]"], 25)

IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

print(f"Q1: {Q1:.3f}")
print(f"Median: {Q2:.3f}")
print(f"Q3: {Q3:.3f}")
print(f"IQR: {IQR:.3f}")
print(f"Upper Bound: {upper_bound:.3f}")
print(f"Lower Bound: {lower_bound:.3f}")

print(f"Outliers count: {len(dfHome[(dfHome["use [kW]"] < lower_bound) | (dfHome["use [kW]"] > upper_bound)])}")
print(f"Outliers percentage: {len(dfHome[(dfHome["use [kW]"] < lower_bound) | (dfHome["use [kW]"] > upper_bound)]) / len(dfHome) * 100:.2f}%")


fig, ax = plt.subplots()
ax.boxplot(dfHome["use [kW]"], vert=True)
ax.set_title("Boxplot of Home Energy use [kw]")
ax.set_ylabel("Energy use [kW]")
plt.show()


```
